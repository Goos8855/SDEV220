```python
#Array sorting Geeksforgeeks problem
class Solution:
    def sort012(self, arr):
        # code here
        one = 0
        two = 0
        zero = 0
        sortArr = []

        for i in arr:
            if i == 1:
                one += 1
            elif i == 2:
                two += 1
            else:
                zero += 1
        
        arr.clear()
        
        for i in range(zero):
            arr.append(0)
        for i in range(one):
            arr.append(1)
        for i in range(two):
            arr.append(2)
            
        return arr

```

```python
#Sorting list using binary search
class Solution:
    def binarysearch(self, arr, k):
    
        lo=0
        hi=len(arr)-1
        result = -1
        while lo <= hi:
            mid = lo+(hi-lo)//2
            if(arr[mid] == k):
                result = mid
                hi = mid-1
            elif arr[mid] < k:
                lo=mid+1
            else:
                hi=mid-1
        
        return result
```
