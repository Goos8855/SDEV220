```python
#4.1
import random
secret = random.randint(1,10)
while(True):
    guess = int(input())
    
    if(guess > secret):
        print("Too high")
    elif(guess < secret):
        print("Too low")
    else:
        print("Just right")
```

```python
#4.2
small = True
green = False

if(small and green):
    print("Pea")
elif(small and not green):
    print("Cherry")
elif(not small and green):
    print("Watermellon")
elif(not small and not green):
    print("Pumpkin")
```

```python
#6.2
guess_me = 7
number = 1
while(True):
    if(number < guess_me):
        print("too low")
        number += 1
    elif(number == guess_me):
        print("found it!")
        break
    elif(number > guess_me):
        print("oops")
        break
        
```
