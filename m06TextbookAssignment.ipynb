```python
import multiprocessing
import time
import random
from datetime import datetime

def printTime():
    wait = random.random()
    time.sleep(wait)
    now = datetime.now()
    print("Time:" + now)

if __name__ == "__main__":
    for i in range(3):
        p = multiprocessing.Process(target=printTime)
        p.start()
```
