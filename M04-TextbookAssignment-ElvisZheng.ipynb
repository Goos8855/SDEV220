```python
#16.8
#The steps for 16.8 looked like they needed the previous parts as well
import sqlalchemy
database = 'books.db'
engine = sqlalchemy.create_engine('sqlite:///books.db')
connection = engine.connect()
metadata = sqlalchemy.MetaData()
book_table = sqlalchemy.Table('book', metadata, autoload_with=engine)
query = sqlalchemy.select(book_table.c.title).order_by(book_table.c.title)

results = connection.execute(query)
for row in results:
    print(row.title)

connection.close()
```
